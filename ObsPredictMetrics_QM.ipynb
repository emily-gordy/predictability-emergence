```python
def bb(forecast1,forecast2,verification,autocorr,nboots):

    nsamps = len(forecast1)
    nsampsgrab = int(np.round(nsamps/autocorr))

    fdiff = np.empty(nboots)
    # f2acc = np.empty(nboots)
    fbss = np.empty(nboots)
    # f2bs = np.empty(nboots)

    for iboot in range(nboots):
        sampsel = np.random.choice(nsamps-autocorr,nsampsgrab,replace=True)
        sampsel = (sampsel[:,np.newaxis]+np.arange(autocorr)).flatten()

        forecast1sel = forecast1[sampsel]
        forecast2sel = forecast2[sampsel]
        verifsel = verification[sampsel]

        fdiff[iboot] = np.mean(np.round(forecast1sel)==verifsel) - np.mean(np.round(forecast2sel)==verifsel)

        f1bs = helpers.brierscore(forecast1sel.squeeze(),verifsel.squeeze())
        f2bs = helpers.brierscore(forecast2sel.squeeze(),verifsel.squeeze())

        # fbss[iboot] = 1-(f1bs/f2bs)
        fbss[iboot] = f1bs-f2bs

    return fdiff,fbss

def expectedcalibrationerror(pred,true,interval):

    confints = np.arange(0,1,interval)

    calibcurve = np.empty(len(confints))
    nsamps = np.empty(len(confints))

    for isel,ii in enumerate(confints):

        MPIsel = (pred>ii) & (pred<(ii+interval))
        calibcurve[isel] = np.mean(true[MPIsel])

        nsamps[isel] = len(true[MPIsel])

    ece = np.nansum(nsamps*np.abs(calibcurve-confints))/len(pred)

    return ece,calibcurve

# def f1score(pred,true):

#     precision = helpers.precision(pred,true,1)
#     recall = helpers.recall(pred,true,1)

#     f1 = 2*(precision*recall)/(precision+recall)

#     return f1

# def f1score(pred,true):

#     TP = (pred==true)&(true==1)
#     FP = (pred!=true)&(pred==1)
#     TN = (pred==true)&(true==0)
#     FN = (pred!=true)&(pred==0)

#     f1 = 2*len(pred[TP])/(2*len(pred[TP]) + len(pred[FP]) + len(pred[FN]))

#     return f1

def f1score(pred, true):
    pred = np.asarray(pred).astype(bool)
    true = np.asarray(true).astype(bool)

    TP = np.sum(pred & true)
    FP = np.sum(pred & ~true)
    FN = np.sum(~pred & true)

    denom = 2 * TP + FP + FN
    if denom == 0:        # all true negatives: no events, none predicted
        return np.nan
    return 2 * TP / denom  # 0 if events/predictions exist but none are hits

# def quantile_correction(allvalpred,qm_correction,bins):

#     edges = np.linspace(0, 1, len(bins)+1)
#     binwidth = edges[1]-edges[0]

#     allvalcorrected = np.empty(len(allvalpred))
#     allvalqm = np.empty(len(bins))

#     for iedge,edge in enumerate(edges[:-1]):
#         predsel = (allvalpred>edge) & (allvalpred<=(edge+binwidth))
#         if ~np.isnan(qm_correction[iedge]):
#             allvalcorrected[predsel] = np.mean(allvalpred[predsel])+qm_correction[iedge]
#             allvalqm[iedge] = np.mean(allvalpred[predsel])
#         else:
#             allvalcorrected[predsel] = np.mean(allvalpred[predsel])
#             allvalqm[iedge] = np.mean(allvalpred[predsel])           

#     return allvalcorrected,allvalqm

def quantile_correction(pred, observedfreq, centres):
    # histogram binning: map each prediction to the observed frequency of its
    # reliability bin on the validation set (bins match eval_reliability.reliability)
    pred = np.asarray(pred, float)
    nbins = len(centres)
    edges = np.linspace(0, 1, nbins+1)
    idx = np.clip(np.searchsorted(edges, pred, side="right") - 1, 0, nbins - 1)

    corrected = observedfreq[idx]
    nodata = np.isnan(corrected)              # empty validation bin: leave uncorrected
    corrected[nodata] = pred[nodata]

    return corrected, idx
```
